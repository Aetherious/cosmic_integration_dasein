### <u>**BlobFuse2 Setup Guide**</u>

##### For mounting an Azure Blob Storage container on Linux (Ubuntu) with BlobFuse2

Source: https://learn.microsoft.com/en-us/azure/storage/blobs/blobfuse2-how-to-deploy

#### **1. Install BlobFuse2**

```python
# Check Linux version
cat /etc/*-release

# Configure the Microsoft package repository
# Example on an Ubuntu 24.04 distribution:
sudo wget https://packages.microsoft.com/config/ubuntu/24.04/packages-microsoft-prod.deb
sudo dpkg -i packages-microsoft-prod.deb
sudo apt-get update
sudo apt-get install libfuse3-dev fuse3
# Change URL to .../ubuntu/16.04/... or .../ubuntu/18.04/... to reference another Ubuntu version.

# Install BlobFuse2
sudo apt-get install blobfuse2
# Similarly, change the package name to blobfuse2=<version> to install specific version.
```

#### **2. Configure BlobFuse2**

Start by creating a config.yml file for setting the mounting configurations.

```python
# ./blobfuse2_config.yml
# Config based on template: https://github.com/Azure/azure-storage-fuse/blob/main/setup/baseConfig.yaml

# Allow non-sudo users to mount filesystem
# sudo nano /etc/<insert>.conf
allow-other: true

# Logger Configs
logging:
  type: syslog
  level: log_warning
  file-path: $HOME/.blobfuse2/blobfuse2.log
```

```python
# Create an empty directory to mount the blob container
mkdir ~/mycontainer
```

#### 3. Mount a Blob Container

```python
sudo blobfuse2 mount ~/mycontainer --config-file=./config.yaml

# Test Deployment
cd ~/mycontainer
mkdir test
echo "hello world" > test/blob.txt
```
